#### 单Agent和多Agent

单Agent的特点  
LangChain内置
- 结构简单，只有一个智能体负责所有任务
- 决策过程集中，不需要协调多个Agent之间的通信
- 适合处理相对简单或专一的任务
- 通常计算资源需求较少
- 响应速度可能更快，因为不需要多Agent间的协调

多Agents的特点   
LangGraph内置
- 系统由多个相互协作的智能体组成
- 各Agent可能有不同的专业领域或功能
- 需要Agent之间的通信和协调机制
- 可以并行处理多个任务
- 系统整体更加复杂但也更加强大
- 具有更好的可扩展性和容错能力

#### 安装 poetry
Mac  
``` cmd
poetry curl -sSL https://install.python-poetry.ory | python3 -
``` 

windows  

powershell 安装命令：  
``` cmd
(Invoke-WebRequest -Uri https://install.python-poetry.org -UseBasicParsing).Content | py -  
``` 

在当前终端临时生效：  
$env:Path += ";$env:APPDATA\Python\Scripts"  

验证：  
``` cmd
poetry --version
``` 

永久添加到系统环境变量（一劳永逸，推荐）  

- 按键盘 Win + S 搜索“环境变量”，选择 “编辑系统环境变量”。
- 点击右下角的 “环境变量” 按钮。
- 在上方的 “用户变量” 列表中，找到并双击 Path。
- 点击 “新建”，添加以下路径（直接复制）：  
```
%APPDATA%\Python\Scripts
```

初始化 poetry  
``` cmd
poetry init  
```

生成依赖锁文件
``` cmd
poetry lock --no-update   
```

安装依赖  
``` cmd
poetry install 
```


确保 Poetry 使用系统默认的 Python
``` cmd
poetry env use py
```

直接运行你的代码 (poetry run 会自动使用虚拟环境，无需手动 activate！)
``` cmd
poetry run python -m src.DingWebHook
```
