### 向量艺术：向量数据库
  

问题被嵌入向量化后进入向量库  
使用向量搜索找到相关文档片段  

- 向量数据库是RAG的关键组件
- 向量数据库中存储了所有的外挂知识
- 向量库本身内置算法对于检索很关键
- 向量库与关系型数据库区别较大   


向量数据库 ≠ 关系数据库

向量数据库
* 向量化存储
* 相似度搜索
* 事务处理

关系数据库
* 结构化存储
* 精确查找
* 事务处理

### 向量艺术：Langchain的向量库的实现

* add_documents - 将文本列表添加到矢量存储中
* delete - 从向量存储中删除文档列表 
* similarity_search - 搜索与给定查询类似的文档
