```python
import asyncio
import edge_tts
import os
import sys
import re
from datetime import datetime, timedelta

# ==========================================
# 📝 在这里粘贴你的带序号的中英混合原始文本
# ==========================================
raw_text = """
1. The little girl faces a life of great hardship. 小女孩面临着艰难的人生
2. She looks so charming. 她看起来真迷人
3. He is charming when he's farming. 他干农活的时候很迷人
4. He's very charming He makes me want to smile at him too. 他是非常迷人的，他让我也想对他微笑
5. The bad weather adds a complication to my vacation. 恶劣的天气给我的假期徒增难题
6. I know in China the number 8 symbolizes prosperity. 我知道在中国数字8代表发达
7. They may be essential for the long-term prosperity. 他们可能对长期繁荣至关重要
8. Happiness dwells in the soul. 快乐取决于心灵
9. The man truly acted coolly. 这个难度真的表现的很淡定
10. I think it's impossible to truly understand someone. 我认为我们无法真正地理解一些人
11. The bridge has five arches. 这座桥有五个拱洞
12. There is an arch above the doorway. 门口上方有一道拱门
"""

def extract_english_sentences(text):
    """
    智能提取纯英文句子：
    1. 自动去除行首的序号（如 "1. ", "10. "），避免 TTS 读出数字。
    2. 自动截断中文字符，只保留纯英文部分（包含英文中的数字）。
    """
    lines = text.strip().split('\n')
    sentences = []
    for line in lines:
        if not line.strip():
            continue
        
        # 步骤 1: 去除行首的数字序号和点（例如 "1. " 或 "12. "）
        clean_line = re.sub(r'^\d+\.\s*', '', line)
        
        # 🌟 步骤 2 (修复): 在字符集中加入了 0-9，防止英文句子中的数字被截断
        match = re.match(r'^([A-Za-z0-9\s\.,!?\'";:-]+)', clean_line)
        if match:
            sentence = match.group(1).strip()
            if sentence:
                sentences.append(sentence)
    return sentences

def merge_mp3_files(file_list, output_filename):
    """智能合并 MP3 文件"""
    print(f"\n🔗 开始合并音频文件...")
    try:
        from pydub import AudioSegment
        print("  ℹ️ 检测到 pydub，正在使用高质量无缝合并模式...")
        combined = AudioSegment.empty()
        for file in file_list:
            audio = AudioSegment.from_mp3(file)
            combined += audio
        combined.export(output_filename, format="mp3", bitrate="320k")
        print(f"  ✅ 高质量合并成功 (320k): {output_filename}")
    except ImportError:
        print("  ⚠️ 未安装 pydub，降级使用纯 Python 二进制合并...")
        try:
            with open(output_filename, 'wb') as outfile:
                for file in file_list:
                    with open(file, 'rb') as infile:
                        outfile.write(infile.read())
            print(f"  ✅ 二进制合并成功: {output_filename}")
        except Exception as e:
            print(f"  ❌ 合并失败: {e}")

async def generate_audio(sentences):
    voices_config = {
        # "adult": {"voice": "en-US-JennyNeural", "rate": "-20%"},       
        "adult": {"voice": "en-US-AriaNeural", "rate": "-20%"},       
        "child": {"voice": "en-US-AnaNeural", "rate": "-20%"}          
    }
    
    output_dir = "edgetts_output"
    os.makedirs(output_dir, exist_ok=True)
    print(f"\n🎙️ 开始生成音频，文件将统一保存在 [{output_dir}] 目录下...")
    print(f"📌 排序方式：大人一句，小孩一句交替进行。")
    print(f"⚡ 语速设置：Adult={voices_config['adult']['rate']}, Child={voices_config['child']['rate']}\n")
    
    file_index = 1  
    
    for i, text in enumerate(sentences):
        for role, config in voices_config.items():
            voice = config["voice"]
            rate = config["rate"]
            
            filename = f"{file_index:03d}_{role}.mp3"
            output_file = os.path.join(output_dir, filename)
            print(f"  [{file_index:02d}] 正在生成 ({role}, 语速:{rate}): {filename}")
            
            try:
                communicate = edge_tts.Communicate(text, voice, rate=rate)
                await communicate.save(output_file)
                print(f"       ✅ 成功: {filename}")
            except Exception as e:
                print(f"       ❌ 失败: {filename} - {e}")
                
            file_index += 1

    print("\n" + "="*50)
    all_files = sorted([os.path.join(output_dir, f) for f in os.listdir(output_dir) if f.endswith('.mp3')])
    
    if all_files:
        # 🌟 修复: 统一使用 %m-%d 格式，确保生成和提示的文件名一致
        tomorrow = datetime.now() + timedelta(days=1)
        date_str = tomorrow.strftime("%m-%d") 
        
        final_filename = f"combined_{date_str}.mp3"
        final_output = os.path.join(output_dir, final_filename)
        
        print(f"📅 最终合并文件命名规则：当前日期 + 1天 -> {date_str}")
        merge_mp3_files(all_files, final_output)
    else:
        print("⚠️ 没有生成任何音频文件，跳过合并步骤。")

# ================= 主程序入口 =================
print("🎧 Edge-TTS 自动化提取与生成程序启动...")

# 1. 自动提取并清洗英文
sentences = extract_english_sentences(raw_text)
print(f"\n🔍 文本解析完成，共提取到 {len(sentences)} 句纯净英文（已自动去除序号、中文，并保留英文数字）：")
for idx, s in enumerate(sentences, 1):
    print(f"   {idx}. {s}")

# 2. 兼容 Jupyter 和 命令行 执行
if 'ipykernel' in sys.modules:
    await generate_audio(sentences)
else:
    asyncio.run(generate_audio(sentences))

print("\n🎉 全部完成！")
print(f"📁 独立分句文件请查看: ./edgetts_output/")

tomorrow = datetime.now() + timedelta(days=1)
final_name = f"combined_{tomorrow.strftime('%m-%d')}.mp3"
print(f"🎵 最终合并的完整音频: ./edgetts_output/{final_name}")
```

```python
import asyncio
import edge_tts
import os
import sys

sentences = [
    "The little girl faces a life of great hardship.",
    "She looks so charming.",
    "He is charming when he's farming.",
    "He's very charming. He makes me want to smile at him too.",
    "The bad weather adds a complication to my vacation.",
    "I know in China the number 8 symbolizes prosperity.",
    "They may be essential for the long-term prosperity.",
    "Happiness dwells in the soul.",
    "The man truly acted coolly.",
    "I think it's impossible to truly understand someone.",
    "The bridge has five arches.",
    "There is an arch above the doorway."
]

def merge_mp3_files(file_list, output_filename):
    """智能合并 MP3 文件"""
    print(f"\n🔗 开始合并音频文件...")
    try:
        # 尝试使用 pydub 进行高质量、无缝的音频拼接
        from pydub import AudioSegment
        print("  ℹ️ 检测到 pydub，正在使用高质量无缝合并模式...")
        combined = AudioSegment.empty()
        for file in file_list:
            audio = AudioSegment.from_mp3(file)
            combined += audio
        # combined.export(output_filename, format="mp3", bitrate="192k")
        combined.export(output_filename, format="wav") 
        print(f"  ✅ 高质量合并成功: {output_filename}")
    except ImportError:
        # 如果没有安装 pydub，则使用纯 Python 二进制合并 (无需任何额外依赖)
        print("  ⚠️ 未安装 pydub，降级使用纯 Python 二进制合并 (无需额外安装)...")
        print("  (💡 提示: 运行 'pip install pydub' 可获得更完美的无缝拼接效果)")
        try:
            with open(output_filename, 'wb') as outfile:
                for file in file_list:
                    with open(file, 'rb') as infile:
                        outfile.write(infile.read())
            print(f"  ✅ 二进制合并成功: {output_filename}")
        except Exception as e:
            print(f"  ❌ 合并失败: {e}")

async def generate_audio():
    # 🌟 升级点：将 voice 和 rate 分开配置，方便独立调整大人和小孩的语速
    voices_config = {
        "adult": {
            "voice": "en-US-JennyNeural", 
            "rate": "-20%"    # 大人语速：正常 (可改为 "+20%" 加快，或 "-20%" 减慢)
        },       
        "child": {
            "voice": "en-US-AnaNeural", 
            "rate": "-20%"   # 小孩语速：稍微快一点，显得更活泼 (可自由修改)
        }          
    }
    
    # 1. 统一放在同一个目录下
    output_dir = "edgetts_output_combined"
    os.makedirs(output_dir, exist_ok=True)
    print(f"\n🎙️ 开始生成音频，文件将统一保存在 [{output_dir}] 目录下...")
    print(f"📌 排序方式：大人一句，小孩一句交替进行。")
    print(f"⚡ 语速设置：Adult={voices_config['adult']['rate']}, Child={voices_config['child']['rate']}\n")
    
    file_index = 1  # 用于全局排序的文件序号
    
    for i, text in enumerate(sentences):
        # 遍历角色配置
        for role, config in voices_config.items():
            voice = config["voice"]
            rate = config["rate"]
            
            # 使用全局序号和角色命名，确保在文件管理器中按顺序排列
            filename = f"{file_index:03d}_{role}.mp3"
            output_file = os.path.join(output_dir, filename)
            print(f"  [{file_index:02d}] 正在生成 ({role}, 语速:{rate}): {filename}")
            
            try:
                # 🌟 核心修改：传入 rate 参数调整语速
                communicate = edge_tts.Communicate(text, voice, rate=rate)
                await communicate.save(output_file)
                print(f"       ✅ 成功: {filename}")
            except Exception as e:
                print(f"       ❌ 失败: {filename} - {e}")
                
            file_index += 1

    # 2. 生成完毕后，执行合并逻辑
    print("\n" + "="*50)
    # 获取目录下所有 mp3 文件，并按文件名排序
    all_files = sorted([os.path.join(output_dir, f) for f in os.listdir(output_dir) if f.endswith('.mp3')])
    
    if all_files:
        final_output = os.path.join(output_dir, "final_combined.mp3")
        merge_mp3_files(all_files, final_output)
    else:
        print("⚠️ 没有生成任何音频文件，跳过合并步骤。")


# 自动兼容 Jupyter 和 普通命令行 的执行逻辑
print("🎧 Edge-TTS 批量生成并合并程序启动...")

if 'ipykernel' in sys.modules:
    await generate_audio()
else:
    asyncio.run(generate_audio())

print("\n🎉 全部完成！")
print(f"📁 独立分句文件请查看: ./edgetts_output_combined/")
print(f"🎵 最终合并的完整音频: ./edgetts_output_combined/final_combined.mp3")
```
