```python
# 起名大师，输出格式为一个数组List
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import BaseOutputParser

import os
from dotenv import load_dotenv

load_dotenv()

# 自定义类
class CommaSeparatedListOutputParser(BaseOutputParser):
    """Parse the output of an LLM call to a comma-separated list."""

    def parse(self, text: str):
        """Parse the output of an LLM Call."""
        # 将文本按逗号分割，并去除每个元素的前后空格
        print(f"Parsing text: {text}")
        return text.strip().split(",")

model = ChatOpenAI(
    model="Qwen/Qwen3.8-27B",  # 模型名称        
    temperature=0,
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
)

prompt = PromptTemplate.from_template("你是一个起名大师，请模仿示例起3个{county}特色的名字，示例：男孩常用名{boy}，女孩常用名{girl}。请返回以逗号分隔的列表形式。仅返回逗号分隔的列表，不要返回其他内容。")
message = prompt.format(county="美国", boy="sam", girl="lucy")
print(message)
res = model.invoke(message)
CommaSeparatedListOutputParser().parse(res.content)
```
