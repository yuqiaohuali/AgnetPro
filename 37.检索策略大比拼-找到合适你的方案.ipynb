### 检索调优：让RAG系统更快更准
---
调优原理：将检索结果通过排序、压缩等算法进行进一步过滤，进一步提高相关性



* 未做检索调优的结果一般依赖于向量库能力
* 对于检索结果的调优可以大幅提升RAG准确度
* 检索调优会产生额外的成本负担
* 上下文压缩：支持单独或管道串联
* 排序：Lost in the middle现象
* 相似性分数：一种精准定位方式
* 混合搜索：不同算法互补得到更好的结果

1. 上下文压缩
    * LLMChainExtractor
    * DocumentCompressorPipeline
2. 排序
    * LongContextRecorder
3. 相似性分数
    * similarity_search_with_score
4. 混合搜索
    * 词法搜索（BMRetriever）
    * 向量相似搜索（FAISS）
