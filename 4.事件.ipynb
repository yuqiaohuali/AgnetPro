```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
load_dotenv()

model = ChatOpenAI(
    model="Qwen/Qwen3.8-27B",  # 模型名称          
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)
question = "langchain是什么？"
```

### Invoke事件
---

```python
model.invoke(question)
```

### Stream事件
---

```python
for chunk in model.stream(question):
    print(chunk.content+"|")
```

### Batch事件
---

```python
model.batch(["langchain的作者是谁？", "langchain的应用场景有哪些？"])
```

### astream_events事件
---

```python
# 1. 将 stream_events 改为 astream_events
async for event in model.astream_events("langchain的竞品有哪些？", version='v2'):
    
    # 2. 使用 .get() 防止 KeyError，因为不是所有事件都有 content
    event_name = event.get("event", "")
    event_data = event.get("data", {})
    
    # 在 v2 版本中，流式内容通常在 data 的 chunk 里
    chunk = event_data.get("chunk", {})
    content = chunk.content if hasattr(chunk, "content") else ""
    
    # 3. 格式化打印
    print(f"event={event_name} | name={event.get('name')} | content={content}")
```

```python
# LLM 的标准事件 - with_structured_output
# 影响LLM输出的结构化数据，通常用于解析JSON或其他结构化格式
from typing import Optional
from pydantic import BaseModel, Field
from langchain_openai import ChatOpenAI
import os

from dotenv import load_dotenv
load_dotenv()

llm = ChatOpenAI(
    model="Qwen/Qwen3.8-27B",  # 模型名称
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)

# Pydantic 模型定义
class Joke(BaseModel):
    """Joke to tell user."""
    setup: str = Field(description="The setup of the joke")
    punchline: str = Field(description="The punchline of the joke")
    rating: Optional[int] = Field(default=None, description="How funny the joke is, on a scale from 1 to 10")

# llm.invoke("给我讲一个关于程序员的笑话")

structured_llm = llm.with_structured_output(Joke)
structured_llm.invoke("给我讲一个关于程序员的笑话")
```

```python
# LLM 的标准事件 - with_structured_output
# 影响LLM输出的结构化数据，通常用于解析JSON或其他结构化格式
from langchain_openai import ChatOpenAI
from typing import Optional, TypedDict, Annotated
import os

from dotenv import load_dotenv
load_dotenv()

llm = ChatOpenAI(
    model="Qwen/Qwen3.8-27B",  # 模型名称
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0.7,
)

# TypedDict 模型定义
class Joke(TypedDict):
    """Joke to tell user."""

    setup: Annotated[str, ..., "The setup of the joke"]
    punchline: Annotated[str, ..., "The punchline of the joke"]
    rating: Annotated[Optional[int], ..., "How funny the joke is, on a scale from 1 to 10"]  # ... means required  
    # rating: Annotated[Optional[int], None, "How funny the joke is, on a scale from 1 to 10"]  # None means optional


structured_llm = llm.with_structured_output(Joke)

for chunk in structured_llm.stream("给我讲一个笑话"):
    print(chunk)
```
