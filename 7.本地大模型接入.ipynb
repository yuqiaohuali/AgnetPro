### 接入本地大模型
---
1. 安装ollama
2. 拉取合适的模型资源
3. 后台运行ollama

#### 安装依赖
---

```python
! pip install langchain-ollama
```

后台开启ollama
- 打开cmd窗口 执行以下命令：ollama serve
- 查看ollama安装的大模型命令：ollama list
- 安装大模型的命令：ollama pull <模型名称>

#### 接入模型
---

```python
from langchain_ollama import ChatOllama
import os

model = ChatOllama(
    model="deepseek-r1:7b", # 模型名称      
    temperature=0
)
model.invoke("介绍下自己")
```

```python
from langchain_ollama import ChatOllama
import os

model = ChatOllama(
    model="qwen2:7b", # 模型名称      
    temperature=0
)
model.invoke("介绍下自己")
```

```python
from langchain_ollama import ChatOllama
import os

model = ChatOllama(
    model="qwen2:7b", # 模型名称      
    temperature=0
)
for chunk in model.stream("介绍下自己"):
    print(chunk.content)
```
