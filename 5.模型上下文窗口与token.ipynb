```python
# 注意目前并不是所有大模型都遵循统一标准
# response_metadata、usage_metadata、request_metadata等属性可能会有所不同
from langchain_deepseek import ChatDeepSeek
import os
from dotenv import load_dotenv
load_dotenv()

model = ChatDeepSeek(
    model="deepseek-ai/DeepSeek-V4.1-Flash",  # 模型名称          
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)
res = model.invoke("介绍下自己")
res.usage_metadata
```

```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
load_dotenv()

model = ChatOpenAI(
    model="Qwen/Qwen3.8-27B",  # 模型名称          
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)
res = model.invoke("介绍下自己")
print(f'OpenAI模型的使用情况: \n{res.response_metadata["token_usage"]}')
```
