整合优化，动态加载三种文件格式,增加了文本切割

```python
from langchain_community.document_loaders import UnstructuredExcelLoader, Docx2txtLoader, PyPDFLoader
from langchain_text_splitters import CharacterTextSplitter

# 定义chatdoc
# defining ChatDoc
class ChatDoc():
    def __init__(self):
        self.doc = None
        self.splitText = [] # 分割后的文本 splited text

    def getFile(self):
        doc = self.doc
        loaders = {
            "docx": Docx2txtLoader,
            "pdf": PyPDFLoader,
            "xlsx": UnstructuredExcelLoader,
        }
        file_extension = doc.split(".")[-1]
        loader_class = loaders.get(file_extension)
        if loader_class:
            try: 
                loader = loader_class(doc)
                text = loader.load()
                return text
            except Exception as e:
                print(f"Error loading {file_extension} files:{e}")
        else:
            print(f"Unsupported file extension: {file_extension}")
            return None

    # 处理文档的函数
    # function to process the document
    def splitSentences(self):
        full_text = self.getFile()  # 获取文档内容 get the document content
        if full_text != None:
            # 对比文档进行分隔
            # split the document
            text_split = CharacterTextSplitter(
                chunk_size=150,
                chunk_overlap=20,
            )
            texts = text_split.split_documents(full_text)
            self.splitText = texts

chat_doc = ChatDoc()
chat_doc.doc = "fake.xlsx"
chat_doc.splitSentences()
print(chat_doc.splitText)
```

```python
# 导入必须的包
# Import the required packages
from langchain_classic.document_loaders import UnstructuredExcelLoader, Docx2txtLoader, PyPDFLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_chroma import Chroma
# 导入聊天所需的模块
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import OpenAIEmbeddings
from langchain_deepseek import ChatDeepSeek

from dotenv import load_dotenv
import os

load_dotenv()

# 使用国产模型
embeddings = OpenAIEmbeddings(
    model="Qwen/Qwen3-Embedding-8B",       # 指定使用的 Embedding 模型名称
    openai_api_key=os.getenv("api_key"),   # 从环境变量读取 API 密钥
    openai_api_base=os.getenv("base_url"),  # 从环境变量读取 API 请求地址（兼容 OpenAI 格式的自定义端点）
    check_embedding_ctx_length=False,   # ★ 直接发原文字符串，第三方端点必须加
    chunk_size=16,                      # ★ 默认一次发 1000 条，网关容易 400/429
    timeout=60,
)

model = ChatDeepSeek(
    model="deepseek-ai/DeepSeek-V4.1-Flash",  # 模型名称          
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)

# 定义chatdoc
# defining ChatDoc
class ChatDoc():
    def __init__(self):
        self.doc = None
        self.splitText = [] # 分割后的文本 splited text
        self.template = [
            ("system", "你是一个处理文档的秘书，你从不说自己是一个大模型或者AI助手，你会根据下面提供的上下文内容来继续回答问题，\n上下文内容\n {context} \n"),
            ("human", "你好！"),
            ("ai", "你好"),
            ("human", "{question}"),
        ]
        self.prompt = ChatPromptTemplate.from_messages(self.template)

    def getFile(self):
        doc = self.doc
        loaders = {
            "docx": Docx2txtLoader,
            "pdf": PyPDFLoader,
            "xlsx": UnstructuredExcelLoader,
        }
        file_extension = doc.split(".")[-1]
        loader_class = loaders.get(file_extension)
        if loader_class:
            try:
                loader = loader_class(doc)
                text = loader.load()
                return text
            except Exception as e:
                print(f"Error loading files: {e}")
        else:
            print(f"Unsupported file extension: {file_extension}")
            return None

    # 处理文档的函数
    # function to process the document
    def splitSentences(self):
        full_text = self.getFile()  # 获取文档内容 get the document content
        if full_text != None:
            # 对文档进行分割
            # split the document
            text_split = CharacterTextSplitter(
                chunk_size = 150,
                chunk_overlap = 20,
            )
            texts = text_split.split_documents(full_text)
            self.splitText = texts

    # 向量化与向量存储
    # Embed
    def embeddingAndVectorDB(self):
        db = Chroma.from_documents(
            documents = self.splitText,
            embedding = embeddings,
        )
        return db

    # 提问并找到相关的文本块，使用最简单的检索器
    # Ask and find relevant text blocks
    def askAndFindFiles(self, question):
        db = self.embeddingAndVectorDB()
        retriever = db.as_retriever(search_type="mmr")
        return retriever.invoke(input=question)

    # 用自然语言和文档聊天
    # Chat with the document using natural language
    def chatWithDoc(self, question):
        _context = ""
        context = self.askAndFindFiles(question)
        for i in context:
            _context += i.page_content

        messages = self.prompt.format_messages(context=_context, question=question)
        strOut = ""
        for chunk in model.stream(messages):
            if chunk.content:          # 跳过思考阶段的空 content
                strOut += chunk.content
                print(strOut)

chat_doc = ChatDoc()
chat_doc.doc = "fake.docx"
chat_doc.splitSentences()
unique_doc = chat_doc.chatWithDoc("乙方的纳税号是啥？")
```
