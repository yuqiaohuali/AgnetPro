```python
from langchain_openai import ChatOpenAI
import os
from dotenv import load_dotenv
load_dotenv()

model = ChatOpenAI(
    model="Qwen/Qwen3.8-27B",  # 模型名称          
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)
res = model.invoke("介绍下自己")
res
```

```python
res = model.invoke("帮我起一个有中国特色的男孩名字")
res.content
```

```python
res = model.invoke("帮我起一个有美国特色的男孩名字")
res.content
```

```python
from langchain_deepseek import ChatDeepSeek
import os
from dotenv import load_dotenv
load_dotenv()

model = ChatDeepSeek(
    model="deepseek-ai/DeepSeek-V4.1-Flash",  # 模型名称          
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)
res = model.invoke("介绍下自己")
res
```

```python
from langchain_anthropic import ChatAnthropic
import os
from dotenv import load_dotenv
load_dotenv()

model = ChatAnthropic(
    model="claude-3-5-sonnet-20241022",  # 模型名称          
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    temperature=0,
)
res = model.invoke("介绍下自己")
res.content
```

```python
from langchain_deepseek import ChatDeepSeek
import os
from dotenv import load_dotenv
load_dotenv()

model = ChatDeepSeek(
    model="deepseek-ai/DeepSeek-V4.1-Flash",  # 模型名称          
    temperature=0.4,
    api_key=os.getenv("api_key"),  # 在环境变量中设置的API Key
    base_url=os.getenv("base_url"), # 在环境变量中设置的API请求地址
    timeout=30,
    max_tokens=200,
    stop="你",
    max_retries=3,
)
res = model.invoke("介绍下自己")
res
```
